#### 0. Imports and settings

```python
import scanpy as sc
import scvelo as scv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import scrublet as scr
import doubletdetection
```

#### 1. Load data

```python
adata_healthy = sc.read('healthy_qc.h5ad')
adata_ulcerated = sc.read('ulcerated_qc.h5ad')
adata_nonulcerated = sc.read('nonulcerated_qc.h5ad')
```

#### 2. Quality control

This step was done for each of the conditions separately. See `Healthy_qc.ipynb`, `Ulcerated_qc.ipynb`, and `Nonulcerated_qc.ipynb`.

#### 3. Merge samples

```python
adata = adata_healthy.concatenate(adata_ulcerated, adata_nonulcerated,
                           batch_key='Condition', 
                           batch_categories=['Healthy', 'Ulcerated', 'Nonulcerated'])
```

#### 4. Save data for the next step

```python
adata.write('all_qc.h5ad')
```
